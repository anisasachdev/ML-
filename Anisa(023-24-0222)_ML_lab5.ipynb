```python
import pandas as pd

df = pd.read_csv("train.csv")

print('Dataset shape:')
print(df.shape)

print('\nFirst 5 Rows:')
display(df.head())

print('Dataset Information:')
df.info()

print('Summary statistics:')
display(df.describe())
```

```python
target = "SalePrice"

numerical_features = [
    "OverallQual",
    "GrLivArea",
    "TotalBsmtSF",
    "1stFlrSF",
    "GarageCars",
    "GarageArea",
    "YearBuilt",
    "FullBath",
    "TotRmsAbvGrd"
]

categorical_features = [
    "Neighborhood",
    "HouseStyle",
    "KitchenQual"
]

print("Dependent Variable:")
print(target)

print("\nNumerical Independent Variables:")
for feature in numerical_features:
    print(feature)

print("\nCategorical Independent Variables:")
for feature in categorical_features:
    print(feature)
```

```python
# This is the Regression problem because the output or the column that we have to predict is 
# sale price which can be any number continuous numerical value.
# As it is a regression model we will use regression metrics to evaluate the model like MAE etc.
```

```python
# part 2
```

```python
X = df[numerical_features+categorical_features]
Y = df[target]

X = pd.get_dummies(X, columns=categorical_features, dtype=int)

print('X shape:', X.shape)
print('Y shape:', Y.shape)

display(X.head)
```

```python
print('Non-numeric columns:')
print(X.dtypes)

print(X.isnull().sum().sum())
print(Y.isnull().sum().sum())
```

```python
# part 3
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, Y_train)
```

```python
import pandas as pd


Y_pred = model.predict(X_test)

comparison = pd.DataFrame({
    'Actual Price' : Y_test.values[:10],
    'Predicted Price': Y_pred[:10]
})

display(comparison)
```

```python
# part 4
```

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

def interpret(row):
    feature = row["Feature"]
    coefficient = row["Coefficient"]

    if "_" in feature:
        category = feature.split("_", 1)[1]

        if coefficient > 0:
            return (
                f"Being in {category} is associated with an increase "
                f"in predicted house price of approximately "
                f"${coefficient:,.2f} compared with the reference category, "
                f"holding other features constant."
            )
        else:
            return (
                f"Being in {category} is associated with a decrease "
                f"in predicted house price of approximately "
                f"${abs(coefficient):,.2f} compared with the reference category, "
                f"holding other features constant."
            )

    else:
        if coefficient > 0 :
            return(
                f"Each additional unit of {feature} is associated with an " 
                f"increase in predicted house price of approximately " 
                f"${coefficient:,.2f}, holding other features constant."
            )
        else:
            return ( f"Each additional unit of {feature} is associated with a " 
                     f"decrease in predicted house price of approximately " 
                     f"${abs(coefficient):,.2f}, holding other features constant." 
            )

coefficients["Interpretation"] = coefficients.apply(interpret, axis=1)

pd.set_option("display.max_colwidth", None)

display(coefficients)
```

```python
# part 5
```

```python
import matplotlib.pyplot as plt

residuals = Y_test - Y_pred

plt.figure(figsize=(8,5))
plt.scatter(Y_pred, residuals)
plt.axhline(y=0, color="red", linestyle="--")
plt.xlabel("Predicted House Price")
plt.ylabel("Residuals")
plt.title("Residuals vs Predicted House Prices")
plt.show()


plt.figure(figsize=(8, 5))
plt.hist(residuals, bins=30)
plt.xlabel("Residuals")
plt.ylabel("Frequency")
plt.title("Histogram of Residuals")
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np


mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, Y_pred)

metrics_table = pd.DataFrame({
    "Metric" : ["MAE", "MSE", "RMSE", "Rsquare"],
    "Value" : [mae, mse, rmse, r2]
})

display(metrics_table)
```

```python
#MAE (Mean Absolute Error)

# MAE tells us the average difference between the actual house prices and the prices predicted by our model. 
# A lower MAE means the predictions are closer to the actual prices.

# MSE (Mean Squared Error)

# MSE measures the average squared error between the actual and predicted values. It gives more importance to large errors because 
# the errors are squared. A lower MSE means better predictions.

### RMSE (Root Mean Squared Error)

# RMSE tells us the typical size of the prediction error while giving more importance to large errors. It is easier to understand 
# than MSE because it is in the same units as the house price. A lower RMSE means better predictions.

# R² (R-squared)

# R² tells us how well the model explains the variation in house prices. Its value is generally between 0 and 1, with a value 
# closer to 1 indicating that the model explains more of the variation in the data.

# Metric I would show to a non-technical stakeholder

# I would show MAE because it is the easiest to understand. For example, if the MAE is $20,000, we can simply say that the model's
# predictions are off by about $20,000 on average. This is much easier for a non-technical person to understand than MSE or R².
```

```python
# part 6
```

```python
Y_train_pred = model.predict(X_train)

train_mae = mean_absolute_error(Y_train, Y_train_pred)
train_mse = mean_squared_error(Y_train, Y_train_pred)
train_rmse = np.sqrt(train_mse)
train_r2 = r2_score(Y_train, Y_train_pred)

comparison_table = pd.DataFrame({
     "Metric": ["MAE", "MSE", "RMSE", "Rsquare"],
    "Training": [train_mae, train_mse, train_rmse, train_r2],
    "Testing": [mae, mse, rmse, r2]
})

display(comparison_table)
```

```python
# There is a minor gap between MAE and RMSE but overall the predicting results are satisfying. The Model has performed well which means
# the model is generalizing the new data well.
```
